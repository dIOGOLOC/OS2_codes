```python
def bandpass_filter(data, fs=250, lowcut=30, highcut=90, order=4):
    """
    Aplica filtro passa-banda Butterworth em um sinal 1D ou 2D (linha = traço).
    
    Parâmetros:
    - data: array (1D ou 2D) com os dados
    - fs: taxa de amostragem (Hz)
    - lowcut: frequência mínima (Hz)
    - highcut: frequência máxima (Hz)
    - order: ordem do filtro (default = 4)

    Retorna:
    - dados filtrados com mesmo shape
    """
    nyq = 0.5 * fs  # Frequência de Nyquist
    low = lowcut / nyq
    high = highcut / nyq
    
    b, a = butter(order, [low, high], btype='band')
    
    # Se for 1D
    if len(data.shape) == 1:
        return filtfilt(b, a, data)
    
    # Se for 2D: aplica traço a traço (por linha)
    elif len(data.shape) == 2:
        return np.array([filtfilt(b, a, trace) for trace in data])
    
    else:
        raise ValueError("Apenas dados 1D ou 2D são suportados.")
```

```python
# Supondo que seu Dataset já está carregado como "ds"
# E tem dimensões: cdp (traces), twt (tempo), data (amplitude)

# Extrai os eixos
cdp = ds.iline.values
twt = ds.samples.values
amplitude = ds.data.values  # shape: (cdp, twt)
amplitude = np.squeeze(amplitude) # remover dimensões de tamanho 1
amplitude = bandpass_filter(amplitude)
```

```python
def bandpass_filter(trace, fs, lowcut, highcut, order=4):
    nyq = 0.5 * fs
    low = lowcut / nyq
    high = highcut / nyq
    b, a = butter(order, [low, high], btype='band')
    return filtfilt(b, a, trace)
```

```python
def filter_dataset_bandpass_ufunc(ds, var="data", fs=250, lowcut=5, highcut=30, order=4):
    """
    Aplica filtro bandpass à variável `var` de um xarray.Dataset 3D (iline, xline, samples)
    sem remover dimensões. Retorna novo Dataset com variável filtrada.
    """
    da = ds[var]

    # Aplica o filtro ao longo da dimensão "samples"
    da_filtered = xr.apply_ufunc(
        bandpass_filter,
        da,
        input_core_dims=[["samples"]],
        output_core_dims=[["samples"]],
        kwargs={"fs": fs, "lowcut": lowcut, "highcut": highcut, "order": order},
        vectorize=True,
        dask="allowed",
        output_dtypes=[da.dtype]
    )

    # Cria cópia do Dataset com a variável filtrada
    ds_filtered = ds.copy()
    ds_filtered[var] = da_filtered

    return ds_filtered
```

```python
ds_filtered = filter_dataset_bandpass_ufunc(ds, var="data", fs=250, lowcut=5, highcut=50)
```
