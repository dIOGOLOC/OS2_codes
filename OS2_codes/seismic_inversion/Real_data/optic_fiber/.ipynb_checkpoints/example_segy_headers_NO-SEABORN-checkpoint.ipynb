# Working with SEG-Y headers

Headers in SEG-Y data are additional meta information asscoiated with each trace. In SEG-Y these are not pooled in a common data block but interleaved with the seismic trace data so we need to do some work to extract it. **segysak** has two helper methods for extracting information from a SEGY file. These are `segy_header_scan` and `segy_header_scrape`. Both of these functions return `pandas.DataFrame` objects containing header or header related information which can be used QC, analysis and plotting.

## Scanning the headers

`segy_header_scan` is primarily designed to help quickly asscertain the byte locations of key header information for loading or converting the full SEG-Y file. It does this by just looking at the first *N* traces (1000 by default) and returns the byte location and statistics related to the file.

```python
def bandpass_filter(trace, fs, lowcut, highcut, order=4):
    nyq = 0.5 * fs
    low = lowcut / nyq
    high = highcut / nyq
    b, a = butter(order, [low, high], btype='band')
    return filtfilt(b, a, trace)
```

```python
def filter_dataset_bandpass_ufunc(ds, var="data", fs=250, lowcut=1, highcut=20, order=4):
    """
    Aplica filtro bandpass à variável `var` de um xarray.Dataset 3D (iline, xline, samples)
    sem remover dimensões. Retorna novo Dataset com variável filtrada.
    """
    da = ds[var]

    # Aplica o filtro ao longo da dimensão "samples"
    da_filtered = xr.apply_ufunc(
        bandpass_filter,
        da,
        input_core_dims=[["samples"]],
        output_core_dims=[["samples"]],
        kwargs={"fs": fs, "lowcut": lowcut, "highcut": highcut, "order": order},
        vectorize=True,
        dask="allowed",
        output_dtypes=[da.dtype]
    )

    # Cria cópia do Dataset com a variável filtrada
    ds_filtered = ds.copy()
    ds_filtered[var] = da_filtered

    return ds_filtered
```

```python
def f(lf, hf,t1,t2):
    
    tiro_1_filtered = filter_dataset_bandpass_ufunc(tiro_1, var="data", fs=250, lowcut=lf, highcut=hf)

    # Create figure and axis
    fig, (ax1,ax2) = plt.subplots(nrows= 1,ncols=2,figsize=(15, 5))
    
    # Normalização dos dados 
    normalized = tiro_1 / np.abs(tiro_1).max(dim='samples')
    
    # Plot each trace (wiggle plot style)
    for i, cdp_val in enumerate(tiro_1.iline.values):
        trace = normalized.sel(iline=cdp_val)
        ax1.plot(trace.data.values.flatten() + i, tiro_1.samples.values, color='black', linewidth=0.5)
   
    # Set Y-axis (samples)
    ax1.invert_yaxis()
    ax1.set_ylabel("Tempo (ms)")
    ax1.set_ylim(t2,t1)
    
    # Set X-axis (cdp index)
    ax1.set_xlabel("Traços (id)")
    ax1.set_xticks(np.arange(0, len(tiro_1.iline), 20))
    ax1.set_xticklabels(tiro_1.iline.values[::20])
    
    ax1.set_title("Dados normalizados")
    
    # ----------------------------------------------------------------------------
    
    # Normalização dos dados 
    normalized = tiro_1_filtered / np.abs(tiro_1_filtered).max(dim='samples')
    
    # Plot each trace (wiggle plot style)
    for i, cdp_val in enumerate(tiro_1_filtered.iline.values):
        trace = normalized.sel(iline=cdp_val)
        ax2.plot(trace.data.values.flatten() + i, tiro_1.samples.values, color='black', linewidth=0.5)
    
    # Set Y-axis (samples)
    ax2.invert_yaxis()
    ax2.set_ylim(t2,t1)
    
    # Set X-axis (cdp index)
    ax2.set_xlabel("Traços (id)")
    ax2.set_xticks(np.arange(0, len(tiro_1_filtered.iline), 20))
    ax2.set_xticklabels(tiro_1_filtered.iline.values[::20])
    
    ax2.set_title("Dados normalizados filtrados")
    
    
    plt.tight_layout()
    plt.show()
    
interactive_plot = interactive(f, lf=(5, 100), hf=(10, 124, 5),t2=(0,5000,100),t1=(0,5000,100))
output = interactive_plot.children[-1]
output.layout.height = '700px'
interactive_plot
```

```python
import segyio
import numpy as np
import xarray as xr
import matplotlib.pyplot as plt
from ipywidgets import interactive

# --- 1. Leitura do SEG-Y e criação do DataArray ---
caminho_segy = '/media/enseadamartel/Expansion/dados_monitor1/estacao258_total.sgy'

with segyio.open(caminho_segy, "r", ignore_geometry=True) as segyfile:
    segyfile.mmap()
    n_tracos = segyfile.tracecount
    tempo = segyfile.samples / 1000  # milissegundos → segundos
    dados = np.stack([segyfile.trace[i] for i in range(n_tracos)], axis=1)

sismico = xr.DataArray(
    dados,
    coords={"tempo": tempo, "traco": np.arange(n_tracos)},
    dims=["tempo", "traco"],
    name="amplitude"
)
```

```python
sismico = sismico.isel(traco=slice(1, 150))
```

```python
sismico.data.max()
```

```python
# --- 2. Função para visualização interativa ---
def plot_divergencia(alpha=2.0, traco_idx=1, t_min=0.0, t_max=10, amp_scale=100):
    if t_max is None or t_max > float(tempo.max()):
        t_max = float(tempo.max())

    
    ganho = xr.DataArray(tempo ** alpha, coords={"tempo": tempo}, dims=["tempo"])
    sismico_corrigido = sismico * ganho

    traco_original = (sismico[:, traco_idx]).values
    traco_corrigido = (sismico_corrigido[:, traco_idx]).values
    ganho_plot = (ganho / ganho.max() * np.max(traco_original)).values

    fig, axs = plt.subplots(1, 3, figsize=(12, 6), sharey=True)

    # Dado original
    axs[0].plot(traco_original, tempo, color="deepskyblue")
    axs[0].set_title("Dado Bruto")
    axs[0].invert_yaxis()
    axs[0].set_xlabel("Amplitude")
    axs[0].set_ylabel("Tempo (s)")
    axs[0].set_xlim(-amp_scale , +amp_scale)
    axs[0].set_ylim(t_max , t_min)
    
    # Função de ganho
    axs[1].plot(ganho_plot, tempo, color="black")
    axs[1].set_title("Função de Ganho\n$GF = t^\\alpha$")
    axs[1].set_xlabel("Ganho (normalizado)")

    # Dado corrigido
    axs[2].plot(traco_corrigido, tempo, color="dodgerblue")
    axs[2].set_title("Dado Corrigido")
    axs[2].set_xlabel("Amplitude")
    axs[2].set_xlim(-amp_scale , +amp_scale)
    axs[2].set_ylim(t_max , t_min)
    
    plt.suptitle(f"Correção da Divergência Esférica (Traço {traco_idx})", fontsize=14)
    plt.tight_layout(rect=[0, 0.03, 1, 0.95])
    plt.show()

# --- 3. Widget interativo ---
interactive_plot = interactive(
    plot_divergencia,
    alpha=(1.5, 2.5, 0.1),
    traco_idx=(0, 140-1, 1),
    t_min=(0.0, 10-0.1, 0.05),
    t_max=(0.05, 10, 0.05),
    amp_scale=(sismico.data.max()/2, sismico.data.max()*2, 1000)
)
output = interactive_plot.children[-1]
output.layout.height = '700px'
interactive_plot
```
